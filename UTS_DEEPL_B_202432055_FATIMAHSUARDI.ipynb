  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>

```python
#202432055_FATIMAHSUARDI
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432055_FATIMAHSUARDI
# Memuat dataset Steam Reviews
df = pd.read_csv('steam_reviews.csv')

# Ambil kolom yang diperlukan: teks review dan label rekomendasi
# Mengubah label 'Recommended'/'Not Recommended' menjadi biner (1 untuk Rekomendasi, 0 untuk Tidak)
df = df.dropna(subset=['review', 'recommendation'])
df['label'] = df['recommendation'].apply(lambda x: 1 if str(x).strip().lower() == 'recommended' else 0)

# Batasi sampel data agar proses training lebih cepat (misal: 3000 baris pertama)
df = df.head(3000)

texts = df['review'].astype(str).values
labels = df['label'].values

# Tokenisasi sederhana berbasis spasi & pembersihan dasar
def simple_tokenize(text):
    import re
    text = re.sub(r'[^a-zA-Z\s]', '', text.lower())
    return text.split()

tokenized_texts = [simple_tokenize(t) for t in texts]

# Membangun Vocabulary (Kamus Kata)
word_counts = Counter([word for doc in tokenized_texts for word in doc])
vocab = {word: idx + 2 for idx, (word, _) in enumerate(word_counts.most_common(5000))}
vocab['<PAD>'] = 0
vocab['<UNK>'] = 1

# Konversi teks ke sequence integer
max_len = 50 # Batas panjang kata per review
def encode_text(tokens, vocab, max_len):
    encoded = [vocab.get(token, vocab['<UNK>']) for token in tokens]
    if len(encoded) < max_len:
        encoded += [vocab['<PAD>']] * (max_len - len(encoded))
    else:
        encoded = encoded[:max_len]
    return encoded

X = np.array([encode_text(tokens, vocab, max_len) for tokens in tokenized_texts])
y = np.array(labels)

# Membagi Data Latih dan Data Uji (80% train, 20% test)
split_idx = int(0.8 * len(X))
X_train, X_test = X[:split_idx], X[split_idx:]
y_train, y_test = y[:split_idx], y[split_idx:]

# Konversi ke PyTorch Tensor & DataLoader
train_data = TensorDataset(torch.tensor(X_train, dtype=torch.long), torch.tensor(y_train, dtype=torch.float32))
train_loader = DataLoader(train_data, batch_size=64, shuffle=True)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432055_FATIMAHSUARDI
class SentimentLSTM(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_dim, output_dim):
        super(SentimentLSTM, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=0)
        self.lstm = nn.LSTM(embedding_dim, hidden_dim, batch_first=True)
        self.fc = nn.Linear(hidden_dim, output_dim)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        embedded = self.embedding(x)
        # LSTM mengembalikan output dan (hidden_state, cell_state)
        lstm_out, (hn, cn) = self.lstm(embedded)
        # Ambil hidden state dari timestep terakhir
        out = self.fc(hn[-1])
        return self.sigmoid(out)

# Inisiasi Hyperparameter Model
VOCAB_SIZE = len(vocab)
EMBEDDING_DIM = 64
HIDDEN_DIM = 32
OUTPUT_DIM = 1

model = SentimentLSTM(VOCAB_SIZE, EMBEDDING_DIM, HIDDEN_DIM, OUTPUT_DIM)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#20432055_FATIMAHSUARDI
epochs = 10
learning_rate = 0.005
criterion = nn.BCELoss() # Binary Cross Entropy untuk klasifikasi biner
optimizer = optim.Adam(model.parameters(), lr=learning_rate)

train_losses = []

model.train()
for epoch in range(epochs):
    epoch_loss = 0
    for inputs, targets in train_loader:
        optimizer.zero_grad()
        outputs = model(inputs).squeeze()
        loss = criterion(outputs, targets)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()

    avg_loss = epoch_loss / len(train_loader)
    train_losses.append(avg_loss)
    print(f'Epoch [{epoch+1}/{epochs}], Loss: {avg_loss:.4f}')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432055_FATIMAHSUARDI
# 1. Plot Grafik Penurunan Loss Training terhadap Epoch
plt.figure(figsize=(10, 4))
plt.plot(range(1, epochs + 1), train_losses, color='#114F6A', linewidth=2, marker='o', label='Training Loss')
plt.title('Grafik Penurunan Loss Selama Pelatihan')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

# 2. Plot Grafik Perbandingan Data Aktual vs Hasil Prediksi Model
# Evaluasi pada data uji (X_test)
model.eval()
with torch.no_grad():
    sample_inputs = torch.tensor(X_test[:30], dtype=torch.long) # Ambil 30 sampel data uji untuk visualisasi yang jelas
    actual_labels = y_test[:30]
    predicted_probs = model(sample_inputs).squeeze().numpy()
    predicted_labels = (predicted_probs >= 0.5).astype(int) # Ambang batas (threshold) 0.5 untuk menentukan kelas sentimen

plt.figure(figsize=(12, 5))
x_axis = np.arange(len(actual_labels))

# Bar chart perbandingan Label Aktual vs Hasil Prediksi Model
plt.bar(x_axis - 0.15, actual_labels, width=0.3, color='#114F6A', label='Data Aktual (0: Negatif, 1: Positif)')
plt.bar(x_axis + 0.15, predicted_probs, width=0.3, color='#1FA4B1', label='Probabilitas Prediksi LSTM')

plt.title('Perbandingan Sentimen Data Aktual vs Hasil Prediksi Model (30 Sampel Data Uji)')
plt.xlabel('Indeks Sampel Data Uji')
plt.ylabel('Label / Probabilitas Sentimen')
plt.xticks(x_axis, [f'S-{i+1}' for i in range(len(actual_labels))], rotation=45, fontsize=8)
plt.axhline(0.5, color='red', linestyle=':', label='Threshold Sentimen (0.5)')
plt.grid(True, linestyle='--', alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```
